```python
#TOP 10 PROFIT / losss
df["Total_Cost_Lakh"] = (
    df["Marketing_Spend_Lakh"]
    + df["Content_Cost_Lakh"]
    + df["Salary_Cost_Lakh"]
    + df["Other_Cost_Lakh"]
)
df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]

print(df[["Product","Revenue_Lakh","Total_Cost_Lakh", "Profit_Lakh"]].head(10))
```

```python
#missing value

print(df.isna().sum())
```

```python
price = 500
quantity = 10
sales = price * quantity

if sales > 4000:

  print("good sales'")
else:
  print("low sales")
```

```python
#DECISON MAKING STRUCTUTRES
#IF STATEMENTS
rating= df["Customer_Rating"].mean()

if rating >= 4:
    print("customer satisfaction is good")
```

```python
#DECISON MAKING STRUCTUTRES
#IF STATEMENTS
rating= df["Revenue_Lakh"].mean()

if rating >= 4:
    print("Revenue is good")
```

```python
#DECISON MAKING
#IF ELSE
revenue=df["Revenue_Lakh"].mean()

if revenue >= 300:
  print("Average revenue is high")
else:
  print("Average revenue is low")
```
